```python
import pandas as pd
import matplotlib.pyplot as plt

nav = pd.read_csv("../data/raw/02_nav_history.csv")

nav['date'] = pd.to_datetime(nav['date'])

plt.figure(figsize=(12,6))

for code in nav['amfi_code'].unique():
    temp = nav[nav['amfi_code'] == code]
    plt.plot(temp['date'], temp['nav'])

plt.title("NAV Trend Analysis (All Schemes)")
plt.xlabel("Date")
plt.ylabel("NAV")
plt.show()
```

```python
aum = pd.read_csv("../data/raw/03_aum_by_fund_house.csv")

aum['date'] = pd.to_datetime(aum['date'])

plt.figure(figsize=(12,6))

for house in aum['fund_house'].unique():
    temp = aum[aum['fund_house'] == house]
    plt.plot(temp['date'], temp['aum_crore'])

plt.title("AUM Trend by Fund House")
plt.xlabel("Date")
plt.ylabel("AUM (Crore)")
plt.show()
```

```python
sip = pd.read_csv("../data/raw/04_monthly_sip_inflows.csv")

plt.figure(figsize=(12,6))
plt.plot(sip['month'], sip['sip_inflow_crore'])

plt.title("Monthly SIP Inflow Trend")
plt.xticks(rotation=90)
plt.show()
```

```python
category = pd.read_csv("../data/raw/05_category_inflows.csv")

category.groupby('category')['net_inflow_crore'].sum().sort_values().plot(kind='barh')

plt.title("Category Wise Net Inflows")
plt.show()
```

```python
txn = pd.read_csv("../data/raw/08_investor_transactions.csv")

txn['gender'].value_counts().plot(kind='pie', autopct='%1.1f%%')

plt.title("Investor Gender Distribution")
plt.ylabel("")
plt.show()
```

```python
txn['age_group'].value_counts().plot(kind='bar')

plt.title("Age Group Distribution")
plt.show()
```

```python
txn['city'].value_counts().head(10).plot(kind='bar')

plt.title("Top 10 Cities")
plt.show()
```

```python
folio = pd.read_csv("../data/raw/06_industry_folio_count.csv")

plt.figure(figsize=(12,6))
plt.plot(folio['month'], folio['total_folios_crore'])

plt.xticks(rotation=90)
plt.title("Industry Folio Growth")
plt.show()
```

```python
portfolio = pd.read_csv("../data/raw/09_portfolio_holdings.csv")

portfolio.groupby('sector')['weight_pct'].sum().sort_values().tail(10).plot(kind='barh')

plt.title("Sector Allocation")
plt.show()
```
